# Análisis Exploratorio: Hechos de Tránsito (2018-2024)

Este notebook forma parte del pipeline de ciencia de datos del proyecto **crash-severity-predictor**.  
El objetivo es explorar los microdatos oficiales de hechos de tránsito proporcionados por el  
Instituto Nacional de Estadística (INE) para el septenio 2018-2024, identificando la estructura,  
calidad y distribución de las variables antes de cualquier transformación.

> **Fuente:** INE — Microdatos de Hechos de Tránsito (2018-2024)  

```python
# --- Importaciones ---------------------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

warnings.filterwarnings('ignore')

# Configuración de visualización
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (12, 5)

print('✓ Librerías cargadas correctamente')
```

#### 1. Carga de Datos

Se cargan los 7 archivos xlsx correspondientes al septenio 2018-2024.  
Cada archivo representa un año de registros de hechos de tránsito.

```python
# -- Carga de datos ---------------------------------------------
RAW_PATH = '../data/raw/ACCIDENTES DE TRANSITO - HECHOS'
YEARS = range(2018, 2025)

frames = []
for year in YEARS:
    path = os.path.join(RAW_PATH, f'hechos-de-transito-ano-{year}.xlsx')
    df_year = pd.read_excel(path)
    df_year['año_carga'] = year
    frames.append(df_year)
    print(f'  {year}: {len(df_year):,} registros cargados')

df = pd.concat(frames, ignore_index=True)
print(f'\n✓ Total registros: {len(df):,}')
```

#### 2. Vista General del Dataset

```python
# -- Dimensiones y tipos de dato ---------------------------------------------
print(f'Dimensiones: {df.shape[0]:,} filas × {df.shape[1]} columnas')
print(f'\nColumnas y tipos de dato:')
print(df.dtypes)
print(f'\nPrimeras 3 filas:')
df.head(3)
```

```python
# -- Columnas por año ---------------------------------------------
for year in YEARS:
    path = os.path.join(RAW_PATH, f'hechos-de-transito-ano-{year}.xlsx')
    cols = pd.read_excel(path, nrows=0).columns.tolist()
    print(f'\n{year}: {cols}')
```

### Hallazgo 1 : Inconsistencia en nombres de columnas

El INE cambió los nombres de columnas entre años:

| Columna estándar | Variantes encontradas | Años |
|---|---|---|
| `num_corre` | `núm_corre`, `Núm_corre`, `num_corre` | 2018-2019, 2020, 2022-2024 |
| `dia_ocu` | `día_ocu`, `Día_ocu`, `dia_ocu` | 2018-2022, 2020, 2023-2024 |
| `dia_sem_ocu` | `día_sem_ocu`, `dia_sem_ocu` | 2018-2022, 2023-2024 |
| `mes_ocu` | `mes_ocu`, `Mes_ooo` | 2018-2019/2021-2024, 2020 |
| `hora_ocu` | `hora_ocu`, `Hora_ocu` | 2018-2019/2021-2024, 2020 |

**Columnas extra:**
- `zona_ciudad` aparece únicamente en 2021 — requiere investigación
- 2020 tiene todas las columnas identificadoras en mayúscula

**Acción en limpieza:** estandarizar todos los nombres a minúscula sin tildes.

```python
# -- Validar si zona_ciudad == zona_ocu en 2021 ---------------------------------------------
df_2021 = pd.read_excel(
    os.path.join(RAW_PATH, 'hechos-de-transito-ano-2021.xlsx')
)

iguales = (df_2021['zona_ocu'] == df_2021['zona_ciudad']).all()
print(f'¿zona_ocu == zona_ciudad en todos los registros? {iguales}')
print(f'\nzona_ocu únicos:     {sorted(df_2021["zona_ocu"].unique())}')
print(f'zona_ciudad únicos:  {sorted(df_2021["zona_ciudad"].unique())}')
```

```python
# -- Ver registros donde difieren ---------------------------------------------
difieren = df_2021[df_2021['zona_ocu'] != df_2021['zona_ciudad']]
print(f'Registros donde difieren: {len(difieren):,} de {len(df_2021):,}')
print(f'\nEjemplos:')
print(difieren[['núm_corre', 'zona_ocu', 'zona_ciudad']].head(10))
```

### Hallazgo 2 : zona_ciudad vs zona_ocu (2021)

`zona_ciudad` y `zona_ocu` **no son la misma variable**.  
En 1,342 de 8,153 registros (16.5%) los valores difieren.  
Cuando `zona_ocu` tiene valor real, `zona_ciudad` registra 99 (ignorado) y viceversa.

**Conclusión:** son dos campos complementarios del mismo concepto geográfico.  
En la limpieza se consolidarán en una sola columna `zona_ocu` tomando  
el valor real de cualquiera de las dos cuando el otro sea 99.

```python
# -- Confirmar que son complementarias ---------------------------------------------
ambas_99 = df_2021[
    (df_2021['zona_ocu'] == 99) & (df_2021['zona_ciudad'] == 99)
]
ambas_valor = df_2021[
    (df_2021['zona_ocu'] != 99) & (df_2021['zona_ciudad'] != 99)
]
print(f'Ambas en 99 (ignorado):      {len(ambas_99):,}')
print(f'Ambas con valor real:         {len(ambas_valor):,}')
print(f'Solo zona_ocu tiene valor:    {len(df_2021[(df_2021["zona_ocu"] != 99) & (df_2021["zona_ciudad"] == 99)]):,}')
print(f'Solo zona_ciudad tiene valor: {len(df_2021[(df_2021["zona_ocu"] == 99) & (df_2021["zona_ciudad"] != 99)]):,}')
```

### Conclusión zona_ciudad (2021)

Los resultados confirman que `zona_ciudad` es completamente redundante:
- Nunca aporta información que `zona_ocu` no tenga (0 registros exclusivos)
- En 1,342 registros `zona_ocu` tiene valor mientras `zona_ciudad` es 99

**Acción en limpieza:** eliminar `zona_ciudad` del año 2021.  
`zona_ocu` es suficiente para el septenio completo.

### Hallazgo 3 : Nulos aparentes por inconsistencia de nombres

Los nulos no representan datos faltantes reales sino columnas duplicadas
por cambios de nomenclatura entre años:

| Columna | Nulos | Causa |
|---|---|---|
| `Núm_corre`, `núm_corre`, `num_corre` | 46,138 / 30,893 / 27,945 | Mismo campo, 3 nombres distintos |
| `Día_ocu`, `día_ocu`, `dia_ocu` | 46,138 / 22,969 / 35,869 | Mismo campo, mayúsculas y tildes |
| `Mes_ocu`, `mes_ocu` | 46,138 / 6,350 | 2020 en mayúscula |
| `zona_ciudad` | 44,335 | Solo existe en 2021 |

**Conclusión:** no hay nulos reales en columnas críticas.  
El problema es de estandarización de nombres, no de calidad de datos.  
**Acción en limpieza:** normalizar todos los nombres a minúscula sin tildes.

```python
# -- Verificar ignorados por variable crítica ----------------------------------------------
IGNORE_CODES = {
    'tipo_veh'    : 99,
    'g_hora_5'    : 4,
    'g_hora'      : 5,
    'zona_ocu'    : 99,
    'tipo_eve'    : 99,
    'marca_veh'   : 999,
    'color_veh'   : 99,
    'modelo_veh'  : 9999,
    'g_modelo_veh': 99,
}

print(f'{"Variable":<15} {"Ignorados":>10} {"% del total":>12}')
print('-' * 40)
for col, code in IGNORE_CODES.items():
    if col in df.columns:
        n = (df[col] == code).sum()
        pct = n / len(df) * 100
        print(f'{col:<15} {n:>10,} {pct:>11.1f}%')
```
